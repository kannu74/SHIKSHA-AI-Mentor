```python
import pandas as pd
import json

# Assuming df1 is already loaded
data1 = [
    {"prompt": row["instruction"], "response": row["response"]}
    for _, row in df1.iterrows()
]
```

```python
df2=df2['conversations']
```

```python
df2.info()
```

```python
df2
```

```python
data2 = []
for convo in df2:
    if isinstance(convo, list) and len(convo) == 2:
        human = next((item["text"] for item in convo if item["user"] == "human"), None)
        gpt = next((item["text"] for item in convo if item["user"] == "gpt"), None)
        if human and gpt:
            data2.append({
                "prompt": human.strip(),
                "response": gpt.strip()
            })
```

```python
df3['dialog'][0]
```

```python
import random
import json

# Assuming data1 and data2 are both lists of dicts like: [{"prompt": ..., "response": ...}, ...]

# Combine them
final_data = data1 + data2

# Shuffle to mix entries
random.shuffle(final_data)
```

```python
with open("human_dataset.json", "w", encoding="utf-8") as f:
    json.dump(final_data, f, ensure_ascii=False, indent=2)
```
